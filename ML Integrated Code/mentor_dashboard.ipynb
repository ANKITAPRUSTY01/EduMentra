```python
import pandas as pd
import joblib

# Load the saved tools
model = joblib.load('risk_model.joblib')
scaler = joblib.load('scaler.joblib')
le = joblib.load('label_encoder.joblib')

def predict_student_risk(student_data):
    """
    Takes a dictionary of student metrics and returns the Risk Level.
    """
    # Convert input to DataFrame
    input_df = pd.DataFrame([student_data])
    
    # Apply the same scaling we used during training
    input_scaled = scaler.transform(input_df)
    
    # Predict
    prediction_numeric = model.predict(input_scaled)
    prediction_label = le.inverse_transform(prediction_numeric)
    
    return prediction_label[0]

# --- EXAMPLE: Connecting a Tutor to a New Mentee ---
# Imagine a new student just submitted their weekly data:
new_mentee_data = {
    'attendance_pct': 52,
    'avg_cgpa': 5.4,
    'backlog_count': 3,
    'dc_count': 4,
    'leaves_taken': 7,
    'late_night_entries': 3,
    'missed_consultations': 5,
    'stress_score': 8,
    'sentiment_score': 0.9
}

risk = predict_student_risk(new_mentee_data)

print(f"--- Mentee Risk Assessment ---")
print(f"Predicted Risk Level: {risk}")

if risk == 'High':
    print("ACTION REQUIRED: Tutor notified to schedule an urgent intervention.")
elif risk == 'Medium':
    print("ACTION: Tutor should check in during the next weekly meeting.")
else:
    print("STATUS: Student is performing well.")
```
