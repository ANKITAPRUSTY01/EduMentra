```python
import pandas as pd
import joblib
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier

# 1. Load the dataset
df = pd.read_csv('mentor_mentee_ml_dataset.csv')

# 2. Data Preprocessing
# We don't need 'roll_no' for the math, so we drop it.
X = df.drop(['roll_no', 'risk_level'], axis=1)
y = df['risk_level']

# Encode the target (High/Medium/Low -> 0, 1, 2)
le = LabelEncoder()
y_encoded = le.fit_transform(y)

# Standardize the features (makes it easier for the model to learn)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Train the Model
# We use Random Forest because it's very robust for tabular data
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_scaled, y_encoded)

# 4. Save everything to use later in your project
joblib.dump(model, 'risk_model.joblib')
joblib.dump(scaler, 'scaler.joblib')
joblib.dump(le, 'label_encoder.joblib')

print("Training Complete! Model and tools saved.")
```
